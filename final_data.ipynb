```python
import pandas as pd
import numpy as np
import json

#%pip install ijson 
```

```python
# Load botwiki JSON data and add a new column to indicate that these are bot accounts
with open('botwiki-2019_tweets.json', 'r') as f:
    data = json.load(f)

# Normalize nested JSON and create DataFrame
df_bots = pd.json_normalize(data)
pd.set_option('display.max_columns', None)
df_bots.head()
df_bots['user.is_human'] = False
```

```python
# Load verified JSON data and add a new column to indicate that these are human accounts
with open('verified-2019_tweets.json', 'r') as f:
    data = json.load(f)

# Normalize nested JSON and create DataFrame
df_verified = pd.json_normalize(data)
pd.set_option('display.max_columns', None)
df_verified['user.is_human'] = True
```

```python
df_verified.head()
```

```python
print(f"Shape of df_verified (rows, columns): {df_verified.shape}")
print(f"Shape of df_bots (rows, columns): {df_bots.shape}")
```

```python
df = pd.concat([df_verified, df_bots], ignore_index=True)
print(f"Shape of combined DataFrame (rows, columns): {df.shape}")
df.columns
```
